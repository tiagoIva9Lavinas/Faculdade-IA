10. Sequência de Fibonacci
*   A lógica utilizada aqui foi, cada número acumula o valor, somando o valor que obteve na soma anterior com o  valor que se manteve do outro número
*   Por exemplo o valor da soma do "num1" irá ser aplicado na soma do "num2", e o valor da soma do "num2" irá ser aplicado na soma do "num1", assim sucessivamente
*   Utilizei o laço de repetição while para repetir uma operação até o final da contagem
*   Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição

```python
num1 = 1
num2 = 1

i = 0
while i < 5:
  num1 = num1 + num2
  print(num1)

  num2 = num2 + num1
  print(num2)

  i = i + 1
```

*   Caso a condição dos laços condicionais sejam atendidas, o atual valor da repetição é atrituído a respectivel variável("i_maior" e "i_menor")

```python
lista = [3,2,6,5,1,4]

i_maior = lista[0]
i_menor = lista[0]
for i in lista:
  if i>i_maior:
    i_maior = i
  if i<i_menor:
    i_menor = i
print("O maior número é", i_maior)
print("O menor número é", i_menor)
```

20. Matriz Transposta
*   Criei uma lista, usando "[]" para colocar os valores dentro dessa lista
*   Utilizei o laço de repetição for para percorrer as listas menores, dentro da lista maior
*   No primeiro for, ele irá rodar a quantidade de colunas da matriz original, resultando na quantiade de linhas da matriz transposta. Assim criando a cada rodada uma nova linha da matriz transposta.
*   O segundo for, ele irá rodar a quantidade de linhas da matriz original, resultando na quantidade de colunas da matriz transposta.

```python
matriz = [[1,2,3],[4,5,6]]
print("Matriz original:")
for n in matriz:
  print(n)

matriz_transposta = []

for coluna_original in range(len(matriz[0])): #roda 3x
  linha_n = [] # cria uma linha nova vazia (isso roda 3x, uma vez por coluna)
  for linha_original in range(len(matriz)): #roda 2x
    linha_n.append(matriz[linha_original][coluna_original])
  matriz_transposta.append(linha_n)

print("Matriz transposta:")
for n in matriz_transposta:
  print(n)
```

21. Busca Binária
*   Usei a função sort() para organizar uma lista em ordem crescente
*   Criei uma variável chamada "posicao_comeco", para armazenar o índice do primeiro elemento da lista
*   Criei uma variável chamada "posicao_final", para armazenar o índice do último elemento da lista
*   Usei a função len() para retornar a quantidade de itens dentro da lista
*   Utilizei o laço de repetição while para repetir uma operação, até a condição definida não ser atendida
*   Criei uma variável "posicao_meio", onde identifica através da divisão com o operador "//", qual a posição central da lista
*   A lógica usada foi, se o valor procurado for maior ou menor que o meio, a posição inicial ou final, irá ser a partir de um índice acima ou abaixo da "posição_meio".

```python
lista = [2,5,7]
lista.sort()
posicao_comeco = 0
posicao_final = len(lista)

valor_encontrar = int(input("Digite o valor que deseja consultar: "))

while(posicao_comeco <= posicao_final):
  posicao_meio = (posicao_comeco + posicao_final) // 2

  if lista[posicao_meio] == valor_encontrar:
    print("Posição encontrada: ", posicao_meio)
    break
  elif lista[posicao_meio] < valor_encontrar:
    posicao_comeco = posicao_meio + 1
  else:
    posicao_final = posicao_meio - 1
```
