20. Matriz Transposta
*   Criei uma lista, usando "[]" para colocar os valores dentro dessa lista
*   Utilizei o laço de repetição for para percorrer as listas menores, dentro da lista maior
*   A lógica utilizada foi: Linhas da matriz original, passam a ser colunas da matriz transposta. Colunas da matriz original, passam a ser linhas na matriz transposta.
*   No primeiro for, ele irá rodar a quantidade de colunas da matriz original, resultando na quantiade de linhas da matriz transposta. Assim criando a cada rodada uma nova linha da matriz transposta.
*   O segundo for, ele irá rodar a quantidade de linhas da matriz original, resultando na quantidade de colunas da matriz transposta.
*   A função "append()" adiciona um valor no final da lista "linha_n", assim adicionando os valores da matriz original, conforme a lógica utilizada.
*   Também utilizei a função "append()", para adicionar as linhas da matriz transposta(linha_n) dentro da lista "matriz_transposta"
*   Utilizei o laço de repetição for para percorrer as duas listas

```python
matriz = [[1,2,3],[4,5,6]]
print("Matriz original:")
for n in matriz:
  print(n)

matriz_transposta = []

for coluna_original in range(len(matriz[0])): #roda 3x
  linha_n = [] # cria uma linha nova vazia (isso roda 3x, uma vez por coluna)
  for linha_original in range(len(matriz)): #roda 2x
    linha_n.append(matriz[linha_original][coluna_original])
  matriz_transposta.append(linha_n)

print("Matriz transposta:")
for n in matriz_transposta:
  print(n)
```

21. Busca Binária
*   Usei a função sort() para organizar uma lista em ordem crescente
*   Criei uma variável chamada "posicao_comeco", para armazenar o índice do primeiro elemento da lista
*   Criei uma variável chamada "posicao_final", para armazenar o índice do último elemento da lista
*   Usei a função len() para retornar a quantidade de itens dentro da lista
*   Utilizei o laço de repetição while para repetir uma operação, até a condição definida não ser atendida
*   Criei uma variável "posicao_meio", onde identifica através da divisão com o operador "//", qual a posição central da lista
*   A lógica usada foi, se o valor procurado for maior ou menor que o meio, a posição inicial ou final, irá ser a partir de um índice acima ou abaixo da "posição_meio".

```python
lista = [2,5,7]
lista.sort()
posicao_comeco = 0
posicao_final = len(lista)

valor_encontrar = int(input("Digite o valor que deseja consultar: "))

while(posicao_comeco <= posicao_final):
  posicao_meio = (posicao_comeco + posicao_final) // 2

  if lista[posicao_meio] == valor_encontrar:
    print("Posição encontrada: ", posicao_meio)
    break
  elif lista[posicao_meio] < valor_encontrar:
    posicao_comeco = posicao_meio + 1
  else:
    posicao_final = posicao_meio - 1
```

22. Criptografia de Cifra de César
*   Através do def, podemos criar uma função em Python. Os parenteses do lado do nome da função servem para colocar os parâmetros.
*   O parâmetro "chave", serve para colocar o tamanho do deslocamento da posição das letras que irão ser criptografadas.
*   Utilizei o laço de repetição for para percorrer a string
*   Utilizei a função isalpha(), para verificar se um caractere ou uma string contém letras ou não. Caso não, ele irá manter igual.
*   Utilizei a função ord(), usada para obter o código Unicode de um caractere
*   Para obter tal letra criptografada, soma-se o código Unicode dessa letra com a "chave".
*   Utilizei a função chr(), para obter um caractere, através do código Unicode, do mesmo.

```python
def crip_cesar(texto, chave):
  texto_cripto = ""
  for l in texto:
    if l.isalpha():
      l_nova = chr(ord(l) + chave)
      texto_cripto = texto_cripto + l_nova
    else:
      texto_cripto = texto_cripto + l
  return "Mensagem criptografada: " + texto_cripto

mensagem = input("Digite a mensagem:")
qtd_chave = int(input("Digite a qunatidade do deslocamento(chave):"))
print(crip_cesar(mensagem, qtd_chave))
```

23. Descriptografar Cifra de César
*   Através do def, podemos criar uma função em Python. Os parenteses do lado do nome da função servem para colocar os parâmetros.
*   O parâmetro "chave", serve para colocar o tamanho do deslocamento da posição das letras que irão ser descriptografadas.
*   Utilizei o laço de repetição for para percorrer a string
*   Utilizei a função ord(), usada para obter o código Unicode de um caractere
*   Para descriptografar tal letra, subtrai o código Unicode dessa letra com a "chave" fornecida.
*   Utilizei a função chr(), para obter um caractere, através do código Unicode, do mesmo.

```python
def descrip_cesar(texto, chave):
  texto_descripto = ""
  for l in texto:
    l_nova = chr(ord(l) - chave)
    texto_descripto = texto_descripto + l_nova
  return "Mensagem criptografada: " + texto_descripto

mensagem = input("Digite a mensagem:")
qtd_chave = int(input("Digite a qunatidade do deslocamento(chave):"))
print(descrip_cesar(mensagem, qtd_chave))
```

25. Classificação de Números
*   Criei listas, usando "[]" para colocar os valores dentro dessas listas
*   Utilizei o laço de repetição for para percorrer a lista
*   Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado
*   Utilizei o laço condicional elif, para definir uma nova condição, caso a condição anterior não tenha sido atendida, else + if
*   Utilizei o laço condicional else, para quando for contrário, à condição, seja realizado algum outro comando
*   A função "append()" adiciona um valor no final da lista.


```python
lista = [1,-3,0,-9,-4,0,7,9,5,-4,2,-1]
lista_positivos = []
lista_negativos = []
lista_zeros = []

for n in lista:
  if n > 0:
    lista_positivos.append(n)
  elif n < 0:
    lista_negativos.append(n)
  else:
    lista_zeros.append(n)

print("Números positivos:", lista_positivos)
print("Números negativos:", lista_negativos)
print("Zeros:", lista_zeros)
```
