```python
#5. Tabuada
#Utilizei o comando int para passar um valor str(texto) para int(número inteiro)
#Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição
#Utilizei o laço de repetição while para repetir uma operação até o final da contagem(variável "i" chegar até 10)
#Criei a variável "resultado" para armazenar o valor da operação de multiplicação
#Comando "print" para imprimir no console, seprando variáveis e textos escritos
#Fiz a soma do valor antigo de "i" com 1, para que aumente o valor de i a cada repetição, para que seja realizada a contagem
num = int(input("Digite um número: "))
i = 1
while i <= 10:
  resultado = num * i
  print(num, "x", i, "=", resultado)
  i = i + 1
```

```python
#7. Número Ímpar ou Par
#Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado
#Utilizei o operador "%" para calcular o resto de alguma divisão, assim calculando qual número é divisível por 2
#Utilizei o laço condicional else, para quando o resultado da operação seja contrário, à condição, seja realizado algum comando
num = int(input("Digite um número: "))
if num % 2 == 0:
  print("Esse número é par!")
else:
  print("Esse número é ímpar")
```

```python
#8. Verificação de Palíndromo
#Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado
#Utilizei o laço condicional else, para quando o resultado da operação seja contrário, à condição, seja realizado algum comando
#Utilizei a ferramenta de fatiamento, usada para inverter e recortar partes de uma string
palavra = input("Digite uma palavra: ")
palavra_pqn = palavra.lower()
palavra_invs = palavra_pqn[::-1]
if palavra_pqn == palavra_invs:
  print("Sim, é um palíndromo!")
else:
  print("Não, não é um plalíndromo")
```

```python
#9. Fatorial de um Número
#Utilizei o laço de repetição while para repetir uma operação até o final da contagem
#Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição, também usado para ser o próximo número na sequência fatorial
##Criei a variável "fatorial" para armazenar e acumular o valor da operação de multiplicação, a cada repetição do laço
num = int(input("Digite um número: "))
i = 1
fatorial = 1
while i <= num:
  fatorial = fatorial * i
  i = i + 1
print(fatorial)
```

```python
#10. Sequência de Fibonacci
#A lógica utilizada aqui foi, cada número acumula o valor, somando o valor que obteve na soma anterior com o  valor que se manteve do outro número
#Por exemplo o valor da soma do "num1" irá ser aplicado na soma do "num2", e o valor da soma do "num2" irá ser aplicado na soma do "num1", assim sucessivamente
#Utilizei o laço de repetição while para repetir uma operação até o final da contagem
#Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição
num1 = 1
num2 = 1

i = 0
while i < 5:
  num1 = num1 + num2
  print(num1)

  num2 = num2 + num1
  print(num2)

  i = i + 1
```

```python
#11. Lista de Números
#Criei uma lista, usando "[]" para colocar os valores dentro dessa lista
#Utilizei "[]" para acessar o valor em determinado índice
#Utilizei o laço de repetição for para percorrer a lista
#Utilizei o laço condicional if, para comparar cada valor se é maior ou menor, que o valor anterior
#Caso a condição dos laços condicionais sejam atendidas, o atual valor da repetição é atrituído a respectivel variável("i_maior" e "i_menor")
lista = [3,2,6,5,1,4]

i_maior = lista[0]
i_menor = lista[0]
for i in lista:
  if i>i_maior:
    i_maior = i
  if i<i_menor:
    i_menor = i
print("O maior número é", i_maior)
print("O menor número é", i_menor)
```

```python
#12. Contagem de Caracteres
#Utilizei uma variável dict, essa armazena valores chave:valor, uma espécie de dois valores em uma só variável
#Utilizei o laço de repetição for para percorrer a string
#Para acessar o valor de uma determinada chave, usamos [] com a chave dentro
#Podemos usar o get também, para acessar/alterar o valor de uma determinada chave, realizando operações, ou só trocando o valor
palavra = input("Digite algo: ")
qtd_caractere = dict()
for letra in palavra:
  qtd_caractere[letra] = qtd_caractere.get(letra, 0) + 1

print(qtd_caractere)
```

```python
#14. Soma dos Elementos de uma Lista
#Usei a função sum() para somar números, ela também recebe como parâmetro um grupo de números(listas)
lista = [3,2,6,5,1,4]

soma = sum(lista)

print("Soma da lista:", soma)
```

```python
#15. Média de uma Lista
#Usei a função sum() para somar números, ela também recebe como parâmetro um grupo de números(listas)
#Usei a função len() para retornar a quantidade de itens dentro da lista
lista = [3,2,6]

media = sum(lista) / len(lista)

print(media)
```

```python
#17. Dicionário de Contagem de Palavras
#Utilizei uma variável dict, essa armazena valores chave:valor, uma espécie de dois valores em uma só variável
#Utilizei o laço de repetição for para percorrer a string
#Para acessar o valor de uma determinada chave, usamos [] com a chave dentro
#Podemos usar o get também, para acessar/alterar o valor de uma determinada chave, realizando operações, ou só trocando o valor
palavra = input("Digite algo: ")
qtd_caractere = dict()
for letra in palavra:
  qtd_caractere[letra] = qtd_caractere.get(letra, 0) + 1

print(qtd_caractere)
```

```python
#18. Número Primo
#Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado
#Utilizei o laço condicional else, para quando for contrário, à condição, seja realizado algum outro comando
#Utilizei o laço condicional elif, para definir uma nova condição, caso a condição anterior não tenha sido atendida, else + if
##Utilizei o operador "%" para calcular o resto de alguma divisão, assim calculando qual número é divisível por 2 e por 3
num = int(input("Digite um número:"))
if num==2 or num==3:
  print(num, "é um número primo!")
elif num%3==0:
  print(num, "não é um número primo!")
elif num%2==0:
  print(num, "não é um número primo!")
else:
  print(num, "é um número primo!")
```
