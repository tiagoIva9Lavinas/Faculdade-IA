20. Matriz Transposta
*   Criei uma lista, usando "[]" para colocar os valores dentro dessa lista
*   Utilizei o laço de repetição for para percorrer as listas menores, dentro da lista maior
*   A lógica utilizada foi: Linhas da matriz original, passam a ser colunas da matriz transposta. Colunas da matriz original, passam a ser linhas na matriz transposta.
*   No primeiro for, ele irá rodar a quantidade de colunas da matriz original, resultando na quantiade de linhas da matriz transposta. Assim criando a cada rodada uma nova linha da matriz transposta.
*   O segundo for, ele irá rodar a quantidade de linhas da matriz original, resultando na quantidade de colunas da matriz transposta.
*   A função "append()" adiciona um valor no final da lista "linha_n", assim adicionando os valores da matriz original, conforme a lógica utilizada.
*   Também utilizei a função "append()", para adicionar as linhas da matriz transposta(linha_n) dentro da lista "matriz_transposta"
*   Utilizei o laço de repetição for para percorrer as duas listas

```python
matriz = [[1,2,3],[4,5,6]]
print("Matriz original:")
for n in matriz:
  print(n)

matriz_transposta = []

for coluna_original in range(len(matriz[0])): #roda 3x
  linha_n = [] # cria uma linha nova vazia (isso roda 3x, uma vez por coluna)
  for linha_original in range(len(matriz)): #roda 2x
    linha_n.append(matriz[linha_original][coluna_original])
  matriz_transposta.append(linha_n)

print("Matriz transposta:")
for n in matriz_transposta:
  print(n)
```

21. Busca Binária
*   Usei a função sort() para organizar uma lista em ordem crescente
*   Criei uma variável chamada "posicao_comeco", para armazenar o índice do primeiro elemento da lista
*   Criei uma variável chamada "posicao_final", para armazenar o índice do último elemento da lista
*   Usei a função len() para retornar a quantidade de itens dentro da lista
*   Utilizei o laço de repetição while para repetir uma operação, até a condição definida não ser atendida
*   Criei uma variável "posicao_meio", onde identifica através da divisão com o operador "//", qual a posição central da lista
*   A lógica usada foi, se o valor procurado for maior ou menor que o meio, a posição inicial ou final, irá ser a partir de um índice acima ou abaixo da "posição_meio".

```python
lista = [2,5,7]
lista.sort()
posicao_comeco = 0
posicao_final = len(lista)

valor_encontrar = int(input("Digite o valor que deseja consultar: "))

while(posicao_comeco <= posicao_final):
  posicao_meio = (posicao_comeco + posicao_final) // 2

  if lista[posicao_meio] == valor_encontrar:
    print("Posição encontrada: ", posicao_meio)
    break
  elif lista[posicao_meio] < valor_encontrar:
    posicao_comeco = posicao_meio + 1
  else:
    posicao_final = posicao_meio - 1
```

22. Criptografia de Cifra de César
*   Através do def, podemos criar uma função em Python. Os parenteses do lado do nome da função servem para colocar os parâmetros.
*   O parâmetro "chave", serve para colocar o tamanho do deslocamento da posição das letras que irão ser criptografadas.
*   Utilizei o laço de repetição for para percorrer a string
*   Utilizei a função isalpha(), para verificar se um caractere ou uma string contém letras ou não. Caso não, ele irá manter igual.
*   Utilizei a função ord(), usada para obter o código Unicode de um caractere
*   Para obter tal letra criptografada, soma-se o código Unicode dessa letra com a "chave".
*   Utilizei a função chr(), para obter um caractere, através do código Unicode, do mesmo.

```python
def crip_cesar(texto, chave):
  texto_cripto = ""
  for l in texto:
    if l.isalpha():
      l_nova = chr(ord(l) + chave)
      texto_cripto = texto_cripto + l_nova
    else:
      texto_cripto = texto_cripto + l
  return "Mensagem criptografada: " + texto_cripto

mensagem = input("Digite a mensagem:")
qtd_chave = int(input("Digite a qunatidade do deslocamento(chave):"))
print(crip_cesar(mensagem, qtd_chave))
```

23. Descriptografar Cifra de César
*   Através do def, podemos criar uma função em Python. Os parenteses do lado do nome da função servem para colocar os parâmetros.
*   O parâmetro "chave", serve para colocar o tamanho do deslocamento da posição das letras que irão ser descriptografadas.
*   Utilizei o laço de repetição for para percorrer a string
*   Utilizei a função ord(), usada para obter o código Unicode de um caractere
*   Para descriptografar tal letra, subtrai o código Unicode dessa letra com a "chave" fornecida.
*   Utilizei a função chr(), para obter um caractere, através do código Unicode, do mesmo.

```python
def descrip_cesar(texto, chave):
  texto_descripto = ""
  for l in texto:
    l_nova = chr(ord(l) - chave)
    texto_descripto = texto_descripto + l_nova
  return "Mensagem criptografada: " + texto_descripto

mensagem = input("Digite a mensagem:")
qtd_chave = int(input("Digite a qunatidade do deslocamento(chave):"))
print(descrip_cesar(mensagem, qtd_chave))
```

*   Usei a função len() para retornar a quantidade de itens dentro da lista
*   Utilizei o laço de repetição for para percorrer a lista
*   O for "posicao" repassa várias vezes a leitura da lista, enquanto o for "elemento" repassa uma vez só. O for "elemento", vai mudando a ordem da lista, com base em uma passada. O for "posicao", irá repassar novamente para o for "elemento", para modificar novamente, a lista modificada anteriormente.
* Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado
* Quando o número anterior for maior que o posterior, os dois irão tocar de posição.

```python
lista = [31, 16, 2, 53, 6]
for posicao in range(len(lista)):
  for elemento in range(len(lista) - 1):
    if lista[elemento] > lista[elemento + 1]:
      lista[elemento], lista[elemento + 1] = lista[elemento + 1], lista[elemento]

print("Lista:", lista)
```

*   No final de cada rodada, será exibido o tabuleiro do jogo.

```python
tabuleiro = ["","","",
             "","","",
             "","",""]

for contagem in range(1, 10):
  tipo_jogador = input("Qual jogador X ou O?")
  if tipo_jogador == "X":
    jogador = int(input("Escolha a posição que deseja jogar (1 a 9):"))
    if tabuleiro[jogador - 1] == "":
      tabuleiro[jogador - 1] = "X"
  elif tipo_jogador == "O":
    jogador = int(input("Escolha a posição que deseja jogar (1 a 9):"))
    if tabuleiro[jogador - 1] == "":
      tabuleiro[jogador - 1] = "O"
  print(tabuleiro)
```

35. Números Fibonacci Até N
*   A lógica utilizada aqui foi, cada número acumula o valor, somando o valor que obteve na soma anterior com o  valor que se manteve do outro número
*   Por exemplo o valor da soma do "num1" irá ser aplicado na soma do "num2", e o valor da soma do "num2" irá ser aplicado na soma do "num1", assim sucessivamente
*   Utilizei o laço de repetição while para repetir uma operação até o final da contagem
*   Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição, até o valor de N

```python
n = int(input("Digite o valor:"))
num1 = 1
num2 = 1

i = 0
while i < n/2:
  num1 = num1 + num2
  print(num1)

  num2 = num2 + num1
  print(num2)

  i = i + 1
```

37. Funções Recursivas
*   Através do def, podemos criar uma função em Python. Os parenteses do lado do nome da função servem para colocar os parâmetros.
*   Utilizei o laço de repetição while para repetir uma operação até o final da contagem
*   Criei uma variável chamada "i" para ser o valor da contagem do nosso laço de repetição, também usado para ser o próximo número na sequência fatorial
*   Criei a variável "fatorial" para armazenar e acumular o valor da operação de multiplicação, a cada repetição do laço

```python
def calc_fatorial(num):
  i = 1
  fatorial = 1
  while i <= num:
    fatorial = fatorial * i
    i = i + 1
  return fatorial

num = int(input("Digite um número:"))
print(calc_fatorial(num))
```

40. Lista de Compras
*   Criei uma lista para armazenar os itens da compra
*   Utilizei o laço de repetição while para repetir uma operação
*   Criei uma variável chamada "op", para o usuário conseguir escolher a opção do serviço desejado.
*   O serviço escolhido pela "op", é feito através dos laços condicionais if e elif
*   A função "append()" adiciona um valor no final da lista.
*   A função "remove()" remove um valor da lista
*   "break" é o que faz o laço de repetição parar

```python
lista_compras = []

while True:
  op = input("1 - Adicionar item \n2 - Remover item \n3 - Listar itens \n4 - Sair")

  if op == "1":
    item = input("Digite o item que deseja adicionar:")
    lista_compras.append(item)
  elif op == "2":
    item = input("Digite o item que deseja remover:")
    if item in lista_compras:
      lista_compras.remove(item)
    else:
      print("Item inválido!")
  elif op == "3":
    print("Lista compras:", lista_compras)
  elif op == "4":
    break
```

41. Remover Duplicatas
*    Criei uma lista "sem_duplicatas", para armazenar a "lista_números" sem números repetidos.
*   Utilizei o laço de repetição for para percorrer a lista
*   Utilizei o laço condicional if, para definir uma condição, caso ela seja atendida, algum comando seja seja realizado. Onde, se o número da "lista_numeros", já existir dentro da lista "sem_duplicatas", significa que esse número é repetido, logo ele não irá ser colocado em "sem_duplicatas". Caso, o número ainda não estiver na lista "sem_dupicatas", ele será adicionado.
*   A função "append()" adiciona um valor no final da lista.

```python
lista_numeros = [1,4,5,3,2,1,4,7,6,3,5,6,3,9,6,2]
sem_duplicatas = []

for n in lista_numeros:
  if n not in sem_duplicatas:
    sem_duplicatas.append(n)

print("A lista sem duplicatas:", sem_duplicatas)
```

48. Ordenação por Inserção
*   O laço de repetição for, é usado para representar cada número, com a sua posição
*   Já o laço de repetição while, é o que faz cada número no for, se comparar com todos os números à esquerda dele, até achar um número que nõ seja maior do que ele, isso tudo em apena uma rodada do for.
*   Caso na comparação dentro do while, o número em "posicao", não é menor que o número antecessor a ele,  o while não irá executar nada, e pasará para próxima rodada do for.
*   No final de cada rodada do while "posicao" diminui 1, pois o elemento está andando para a esquerda, logo a posição dele irá diminuir, até achar a posição correta.

```python
lista = [25,3,13,8,5]

for posicao in range(1, len(lista)):
  while posicao > 0 and lista[posicao] < lista[posicao - 1]:
    lista[posicao], lista[posicao - 1] = lista[posicao - 1], lista[posicao]
    posicao = posicao - 1

print("Lista orenada:", lista)
```
