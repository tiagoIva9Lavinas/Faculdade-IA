19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado).

```python
import numpy as np

array1 = np.random.randint(1,11,20)
array2 = np.random.randint(1,11,20)

produto_escalar = np.dot(array1,array2)

norma1 = np.sqrt(np.sum(array1 ** 2))
norma2 = np.sqrt(np.sum(array2 ** 2))

cos = produto_escalar / (norma1 * norma2)

angl_rad = np.arccos(np.clip(cos, -1, 1))

print("Cosseno:", cos)
print("Ângulo radianos:", angl_rad)
print("Ângulo em graus:", np.degrees(angl_rad))
```

1. Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π.

```python
import scipy.integrate as integrate
import numpy as np

def f(x):
  return (x ** 2) * np.sin(x)

resultado, erro = integrate.quad(f,0,np.pi)

print(resultado)
```

2. Resolva a equação diferencial dy / dx = x - y, com y(0) = 1, no intervalo de 0 a 10.

```python
import scipy.integrate as integrate
import numpy as np

def eq_diferencial(x,y):
  return x - y

solucao = integrate.solve_ivp(eq_diferencial, [0,10], [1])
print("Solução:", solucao.y[0][-1])
```

4. Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x + 1.

```python
import scipy.optimize as optimize

def f(x):
  return -x**2 + 4*x + 1

def func_negativa(x):
  return -f(x)

resultado = optimize.minimize(func_negativa,0)
print("Mínimo encontrado em x:", resultado.x)
```

16. Resolva a equação diferencial de segunda ordem (d²y / dx²) +y = 0, com condições
iniciais y(0) = 1 e y′(0) = 0.

```python
from scipy.integrate import solve_ivp
import numpy as np

def eq_diferencial(x, y_v):
  y, dy_dx = y_v
  return [dy_dx, -y]

solucao = solve_ivp(eq_diferencial, [0,5], [1,0], t_eval = np.linspace(0,5,100))

print("Solução:", solucao.y[0][-1])
```

10. Plote três gráficos de linha em um único gráfico, cada um com uma cor e
estilo de linha diferentes.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)

plt.plot(x, y1, color='blue', linestyle='-')
plt.plot(x, y2, color='green', linestyle=':')
plt.plot(x, y3, color='red', linestyle='--')

plt.title("Grafico com 3 gráficos")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.legend()
plt.show()
```

11. Crie um gráfico de barras empilhadas para três categorias com três grupos
de valores.

```python
import matplotlib.pyplot as plt
import numpy as np

categorias = ['A', 'B', 'C']
valores1 = [4, 7, 2]
valores2 = [1, 6, 3]
valores3 = [2, 6, 9]

plt.bar(categorias, valores1, label= "Grupo1")
plt.bar(categorias, valores2, bottom=valores1, label= "Grupo2")
plt.bar(categorias, valores3, bottom=valores2, label= "Grupo3")

plt.title("Gráfico de Barras Empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.legend()
plt.show()
```

12. Plote um gráfico de linha com uma função quadrática e ajuste os limites
do eixo para focar nos valores x= −5 a x=5.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)
y = x ** 2

plt.plot(x, y)
plt.title("Gráfico de Linha - Função Quadrática")
plt.xlim(-5, 5)
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True)
plt.show()
```

13. Gere um gráfico de dispersão 3D para 100 pontos aleatórios utilizando
Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

figura = plt.figure()
ax = figura.add_subplot(projection='3d')

ax.scatter(x, y, z)
ax.set_title("Gráfico de Dispersão em 3D")
ax.set_xlabel("Eixo X")
ax.set_ylabel("Eixo Y")
ax.set_zlabel("Eixo Z")
ax.show()
```

16. Crie um gráfico de linha com títulos para cada eixo e adicione um rótulo
de texto no ponto x = 5.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10 , 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Gráfico de linha - Exercício 16")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.text(5, np.sin(5), "x = 5", color='red')
plt.show()
```

18. Crie um gráfico de barras com valores positivos e negativos e use diferentes
cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt
import numpy as np

x = ["Categoria 1", "Categoria 2", "Categoria 3", "Categoria 4", "Categoria 5"]
y = [5,-3,7,9,-2]
cores = ['green' if v >= 0 else 'red' for v in y]

plt.bar(x,y,color=cores)
plt.title("Gráfico de Barras Exercício 4")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

19. Plote um gráfico de linha com duas séries e adicione uma legenda para
diferenciá-las.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin")
plt.plot(x, y2, label="cos")

plt.title("Grafico com duas série")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.legend()
plt.show()
```

20. Adicione uma grade com espaçamento personalizado a um gráfico de linha
para y = x²

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = x**2

plt.plot(x, y)

plt.xticks(np.arange(-10,11,2))
plt.yticks(np.arange(0, 101, 10))

plt.title("Grafico de Linha - Exercício 20")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.grid(True)
plt.show()
```
